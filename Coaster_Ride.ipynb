```python
df = df[['coaster_name', 
    #'Length', 'Speed', 
    'Location', 'Status', 
    #'Opening date',
       #'Type', 
       'Manufacturer', 
       #'Height restriction', 'Model', 'Height',
      # 'Inversions', 'Lift/launch system', 'Cost', 'Trains', 'Park section',
      #'Duration', 'Capacity', 'G-force', 'Designer', 'Max vertical angle',
       #'Drop', 'Soft opening date', 'Fast Lane available', 'Replaced',
       #'Track layout', 'Fastrack available', 'Soft opening date.1',
       #'Closing date', 
      # 'Opened', 
       #'Replaced by', 'Website',
       #'Flash Pass Available', 'Must transfer from wheelchair', 'Theme',
       #'Single rider line available', 'Restraint Style',
      # 'Flash Pass available', 'Acceleration', 'Restraints', 'Name',
       'year_introduced', 
       'latitude', 'longitude', 
       'Type_Main',
       'opening_date_clean', 
       #'speed1', 'speed2', 'speed1_value', 'speed1_unit',
       'speed_mph', 
       #'height_value', 'height_unit', 
       'height_ft',
       'Inversions_clean', 'Gforce_clean']].copy()
```

```python
# Convert the 'opening_date_clean' column to datetime
df['opening_date_clean'] = pd.to_datetime(df['opening_date_clean'])
```

```python
df = df.rename(columns={'coaster_name': 'Coaster_Name',
                   'year_introduced': 'Year_Introduced',
                   'latitude': 'Latitude',
                   'longitude': 'Longitude',
                   'opening_date_clean': 'Opening_Date',
                   'speed_mph': 'Speed_MPH',
                   'height_ft': 'Height_FT',
                   'Inversions_clean': 'Inversions',
                   'Gforce_clean': 'Gforce'})
```

```python
#checking for 
df.isna().sum()
```

```python
#locating duplicated data
df.loc[df.duplicated()]
```

```python
df.loc[df.duplicated(subset=['Coaster_Name'])].head(5)
```

```python
# Query for a specific coaster for duplicate checking
df.query('Coaster_Name == "Crystal Beach Cyclone"')
```

```python
#Resetting dataframe index
df = df.loc[~df.duplicated(subset=['Coaster_Name', 'Opening_Date'])].reset_index(drop=True)
```

```python
df.shape
```

```python
#ploting bar chart
ax = df['Year_Introduced'].value_counts()\
   .head(10) \
   .plot(kind='bar', title= 'Top 10 Years Coasters Introduced')
ax.set_xlabel('Year Introduced')
ax.set_ylabel('Count')
plt.show()
```

```python
#plotting kde for Coster speed
ax = df['Speed_MPH'].plot(kind='kde', title='Distribution of Coaster Speeds (MPH)')
ax.set_xlabel('Speed (MPH)')
ax.set_ylabel('Count')
plt.show()
```

```python
#plotting histogram for Coster speed
ax = df['Speed_MPH'].plot(kind='hist', bins=20, title='Distribution of Coaster Speeds (MPH)')
ax.set_xlabel('Speed (MPH)')
ax.set_ylabel('Count')
plt.show()
```

```python
#plotting scatter to show relationship between speed and height 
df.plot(kind="scatter", x='Speed_MPH', 
        y="Height_FT",
        title= 'Coaster Speed Vs Height')
plt.show()
```

```python
#plotting scatter plot using Seaborn for Coster speed, year and height
ax = sns.scatterplot(data=df, x='Speed_MPH', y='Height_FT', hue='Year_Introduced',)
ax.set_title('Coaster Speed Vs Height')
plt.show()
```

```python
#plotting pairplot plot using Seaborn for Coster speed, year and height, inversions and Gforce
ax = sns.pairplot(df, vars=['Year_Introduced', 'Speed_MPH', 'Height_FT', 'Inversions', 'Gforce'], hue='Type_Main')
plt.show()
```

```python
#Correlation for Coster speed, year and height, inversions and Gforce
df_corr = df[['Year_Introduced', 'Speed_MPH', 'Height_FT', 'Inversions', 'Gforce']].dropna().corr()
```

```python
#Seaborn heatmap for correlation 
sns.heatmap(df_corr, annot=True)
```

```python
#plotting barh for Location speed to determine the mean and average speed 
ax = df.query('Location != "Other"')\
   .groupby('Location')['Speed_MPH']\
   .agg(['mean', 'count'])\
   .query('count >= 10')\
   .sort_values('mean')['mean']\
   .plot(kind='barh', figsize=(10,8), title= 'Average Coster Speed by Location')
ax.set_xlabel('Average Coaster Speed')
plt.show()
```
